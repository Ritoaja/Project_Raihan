```python
"""
01 - EKSPLORASI LIBRARY PYTHON UNTUK PROBABILITAS
Setiap bagian: nama library, fungsi utama, contoh pemakaian.
"""
import math, itertools, random, statistics
from fractions import Fraction
from collections import Counter

import numpy as np
import pandas as pd
import sympy as sp
from scipy import stats
import matplotlib.pyplot as plt

def judul(t): print("\n" + "=" * 70 + f"\n{t}\n" + "=" * 70)

# ---------------------------------------------------------------- 1. math
judul("1. math  -> faktorial, permutasi, kombinasi, fungsi dasar")
print("math.factorial(5)  =", math.factorial(5))
print("math.comb(10, 3)   =", math.comb(10, 3), " (kombinasi C(10,3))")
print("math.perm(10, 3)   =", math.perm(10, 3), " (permutasi P(10,3))")
print("math.exp(-2)       =", round(math.exp(-2), 4))

# ----------------------------------------------------------- 2. fractions
judul("2. fractions -> probabilitas dalam bentuk pecahan eksak")
p = Fraction(1, 6) + Fraction(1, 6)
print("P(muncul 1 atau 2 pada dadu) =", p, "=", float(p))

# ---------------------------------------------------------- 3. itertools
judul("3. itertools -> enumerasi ruang sampel")
ruang = list(itertools.product(range(1, 7), repeat=2))
print("Jumlah titik sampel 2 dadu :", len(ruang))
print("Permutasi 'ABC'            :", ["".join(x) for x in itertools.permutations("ABC")])
print("Kombinasi 2 dari 'ABCD'    :", ["".join(x) for x in itertools.combinations("ABCD", 2)])

# ------------------------------------------------------------- 4. random
judul("4. random -> simulasi percobaan acak sederhana")
random.seed(1)
print("Lempar koin 10x :", [random.choice("HT") for _ in range(10)])
print("random.sample   :", random.sample(range(1, 50), 6), "(undian 6 dari 49)")

# ----------------------------------------------------------- 5. statistics
judul("5. statistics -> mean, median, modus, varians, stdev")
data = [2, 3, 3, 5, 7, 8, 8, 8, 10]
print("mean =", round(statistics.mean(data), 3), "| median =", statistics.median(data),
      "| modus =", statistics.mode(data))
print("variance =", round(statistics.variance(data), 3), "| stdev =", round(statistics.stdev(data), 3))

# ----------------------------------------------------------- 6. collections
judul("6. collections.Counter -> hitung frekuensi hasil simulasi")
rng = np.random.default_rng(42)
c = Counter(rng.integers(1, 7, 6000))
print(dict(sorted(c.items())))

# -------------------------------------------------------------- 7. numpy
judul("7. numpy -> generator bilangan acak & operasi array")
print("rng.binomial(10, .5, 5) :", rng.binomial(10, .5, 5))
print("rng.poisson(3, 5)       :", rng.poisson(3, 5))
print("rng.normal(0, 1, 3)     :", np.round(rng.normal(0, 1, 3), 3))
print("np.mean / np.var (10000 sampel uniform):",
      round(rng.random(10000).mean(), 4), round(rng.random(10000).var(), 4))

# -------------------------------------------------------------- 8. scipy.stats
judul("8. scipy.stats -> distribusi (pmf/pdf, cdf, sf, ppf, mean, var, rvs)")
X = stats.binom(n=10, p=0.5)
print("pmf(5)=", round(X.pmf(5), 4), "| cdf(5)=", round(X.cdf(5), 4),
      "| sf(5)=", round(X.sf(5), 4), "| ppf(.9)=", X.ppf(.9))
print("mean =", X.mean(), "| var =", X.var(), "| std =", round(X.std(), 4))
print("rvs(5) =", X.rvs(5, random_state=1))

# -------------------------------------------------------------- 9. sympy
judul("9. sympy -> perhitungan probabilitas simbolik & kalkulus")
x, lam = sp.symbols("x lambda", positive=True)
pdf = lam * sp.exp(-lam * x)
print("Cek total pdf eksponensial :", sp.integrate(pdf, (x, 0, sp.oo)))
print("E[X] simbolik              :", sp.integrate(x * pdf, (x, 0, sp.oo)))
print("Var[X] simbolik            :", sp.simplify(sp.integrate(x**2 * pdf, (x, 0, sp.oo))
                                                   - sp.integrate(x * pdf, (x, 0, sp.oo))**2))
print("Binomial(3, p) P(X=2)      :", sp.expand(sp.binomial(3, 2) * sp.Symbol("p")**2 * (1 - sp.Symbol("p"))))

# -------------------------------------------------------------- 10. pandas
judul("10. pandas -> tabel frekuensi & tabel kontingensi")
df = pd.DataFrame({"dadu1": rng.integers(1, 7, 1000), "dadu2": rng.integers(1, 7, 1000)})
df["jumlah"] = df.dadu1 + df.dadu2
print(df["jumlah"].value_counts(normalize=True).sort_index().round(3).head(6))
print(pd.crosstab(df.dadu1 > 3, df.dadu2 > 3, normalize=True).round(3))

# -------------------------------------------------------------- 11. matplotlib
judul("11. matplotlib -> visualisasi (histogram + kurva teoritis)")
s = rng.normal(0, 1, 5000)
plt.hist(s, bins=40, density=True, alpha=.6, label="Simulasi")
xs = np.linspace(-4, 4, 200)
plt.plot(xs, stats.norm.pdf(xs), "r", label="pdf N(0,1)")
plt.legend(); plt.title("Contoh matplotlib + scipy.stats"); plt.savefig("g01_library_demo.png", dpi=120)
plt.show(); print("Grafik disimpan: g01_library_demo.png")

# -------------------------------------------------------------- ringkasan
judul("RINGKASAN")
tabel = pd.DataFrame({
    "Library": ["math", "fractions", "itertools", "random", "statistics", "collections",
                "numpy", "scipy.stats", "sympy", "pandas", "matplotlib"],
    "Fungsi utama": ["comb, perm, factorial, exp", "Fraction", "product, permutations, combinations",
                     "choice, sample, seed", "mean, variance, mode", "Counter",
                     "random.default_rng, mean, var", "pmf/pdf, cdf, ppf, rvs, fit",
                     "integrate, symbols, simplify", "crosstab, value_counts", "hist, plot, bar"],
    "Kegunaan": ["hitung kombinatorik", "peluang eksak", "ruang sampel", "simulasi dasar",
                 "statistik deskriptif", "frekuensi", "simulasi vektor cepat",
                 "distribusi probabilitas", "turunan/integral simbolik", "tabel data",
                 "grafik"]})
print(tabel.to_string(index=False))
```

```python
"""
02 - PERMASALAHAN PROBABILITAS
Tiap soal: (a) hitung secara matematis, (b) hitung/enumerasi dengan Python,
(c) simulasi Monte Carlo bila memungkinkan, lalu dibandingkan + interpretasi.
"""
import math, itertools
import numpy as np
import pandas as pd
import sympy as sp
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
N = 200_000   # jumlah simulasi

def judul(t): print("\n" + "=" * 70 + f"\n{t}\n" + "=" * 70)

# ======================================================================
judul("SOAL 1 - Dua dadu: P(jumlah = 7) [ruang sampel & Monte Carlo]")
ruang = list(itertools.product(range(1, 7), repeat=2))
p_enum = sum(1 for a, b in ruang if a + b == 7) / len(ruang)
d = rng.integers(1, 7, (N, 2)).sum(axis=1)
print(f"Matematis  : 6/36        = {6/36:.4f}")
print(f"Enumerasi  : {p_enum:.4f}")
print(f"Simulasi   : {(d == 7).mean():.4f}")
print("Interpretasi: jumlah 7 punya kombinasi terbanyak (6 dari 36), jadi paling sering muncul.")

# ======================================================================
judul("SOAL 2 - Kombinatorik: tim 3 orang dari 10 mhs (4 perempuan). P(tepat 2 perempuan)")
p = math.comb(4, 2) * math.comb(6, 1) / math.comb(10, 3)
sim = np.mean([sum(rng.choice(10, 3, replace=False) < 4) == 2 for _ in range(50_000)])
print(f"Matematis  : C(4,2)*C(6,1)/C(10,3) = {p:.4f}")
print(f"Simulasi   : {sim:.4f}")
print("Interpretasi: sekitar 30% peluang tim terdiri dari tepat 2 perempuan.")

# ======================================================================
judul("SOAL 3 - Aturan penjumlahan: P(Python)=.6, P(Java)=.4, P(keduanya)=.25")
pA, pB, pAB = .6, .4, .25
print(f"P(A atau B)           = {pA + pB - pAB:.2f}")
print(f"P(tidak keduanya)     = {1 - (pA + pB - pAB):.2f}")
print(f"P(hanya Python)       = {pA - pAB:.2f}")
mhs = rng.random((N, 2))
A = mhs[:, 0] < .6
B = np.where(A, rng.random(N) < pAB / pA, rng.random(N) < (pB - pAB) / (1 - pA))
print(f"Simulasi P(A atau B)  = {(A | B).mean():.4f} | P(A dan B) = {(A & B).mean():.4f}")
print("Interpretasi: 75% mahasiswa memakai minimal satu bahasa; 25% tidak memakai keduanya.")

# ======================================================================
judul("SOAL 4 - Probabilitas bersyarat & independensi (tabel kontingensi)")
tab = pd.DataFrame({"Lulus": [45, 25], "Tidak": [5, 25]},
                   index=["Rajin Latihan", "Jarang Latihan"])
tot = tab.values.sum()
print(tab, "\n")
P_lulus = tab["Lulus"].sum() / tot
P_rajin = tab.loc["Rajin Latihan"].sum() / tot
P_l_r = tab.loc["Rajin Latihan", "Lulus"] / tab.loc["Rajin Latihan"].sum()
print(f"P(Lulus)            = {P_lulus:.3f}")
print(f"P(Lulus | Rajin)    = {P_l_r:.3f}")
print(f"P(Lulus | Jarang)   = {tab.loc['Jarang Latihan','Lulus']/tab.loc['Jarang Latihan'].sum():.3f}")
print(f"Independen? P(L)*P(R)={P_lulus*P_rajin:.3f} vs P(L,R)={tab.loc['Rajin Latihan','Lulus']/tot:.3f}")
print("Interpretasi: P(Lulus|Rajin)=0.9 >> P(Lulus)=0.7, maka lulus & rajin latihan TIDAK independen.")

# ======================================================================
judul("SOAL 5 - Teorema Bayes: Filter spam")
P_spam, P_k_spam, P_k_ham = .30, .80, .10
P_k = P_k_spam * P_spam + P_k_ham * (1 - P_spam)
post = P_k_spam * P_spam / P_k
print(f"P(kata 'gratis')           = {P_k:.4f}  (total probabilitas)")
print(f"P(spam | kata 'gratis')    = {post:.4f}")
email_spam = rng.random(N) < P_spam
ada_kata = np.where(email_spam, rng.random(N) < P_k_spam, rng.random(N) < P_k_ham)
print(f"Simulasi                   = {email_spam[ada_kata].mean():.4f}")
print("Interpretasi: prior spam 30% naik jadi ~77% setelah melihat kata 'gratis'.")

# ======================================================================
judul("SOAL 6 - Bayes: Antivirus (base rate fallacy)")
prev, sens, fpr = .01, .95, .05
post = sens * prev / (sens * prev + fpr * (1 - prev))
print(f"P(malware | alarm) = {post:.4f}")
inf = rng.random(N) < prev
alarm = np.where(inf, rng.random(N) < sens, rng.random(N) < fpr)
print(f"Simulasi           = {inf[alarm].mean():.4f}")
print("Interpretasi: walau akurasi 95%, hanya ~16% alarm benar-benar malware karena malware sangat langka.")

# ---- versi simbolik dengan sympy
pr, se, fp = sp.symbols("prev sens fpr")
bayes = se * pr / (se * pr + fp * (1 - pr))
print("Rumus simbolik   :", sp.simplify(bayes))
print("Substitusi angka :", float(bayes.subs({pr: .01, se: .95, fp: .05})))

# ======================================================================
judul("SOAL 7 - Total probabilitas + Bayes: 3 server")
w = np.array([.5, .3, .2]); err = np.array([.01, .02, .05])
P_err = (w * err).sum()
print(f"P(error)            = {P_err:.4f}")
for i, post in enumerate(w * err / P_err, 1):
    print(f"P(Server {i} | error) = {post:.4f}")
s = rng.choice(3, N, p=w); e = rng.random(N) < err[s]
print("Simulasi            =", np.round([(s[e] == i).mean() for i in range(3)], 4))
print("Interpretasi: Server 3 hanya 20% trafik, tapi menyumbang ~48% error karena error rate tertinggi.")

# ======================================================================
judul("SOAL 8 - Monty Hall (simulasi)")
pintu = rng.integers(0, 3, N); pilih = rng.integers(0, 3, N)
print(f"Tetap   : {(pintu == pilih).mean():.4f}  (teori 1/3)")
print(f"Berganti: {(pintu != pilih).mean():.4f}  (teori 2/3)")
print("Interpretasi: berganti pintu menggandakan peluang menang.")

# ======================================================================
judul("SOAL 9 - Birthday paradox (cocok untuk hashing collision)")
def p_sama(n): return 1 - np.prod([(365 - i) / 365 for i in range(n)])
print(f"Eksak n=23 : {p_sama(23):.4f}")
sim = np.mean([len(set(rng.integers(0, 365, 23))) < 23 for _ in range(30_000)])
print(f"Simulasi   : {sim:.4f}")
print("Interpretasi: cukup 23 orang untuk peluang >50% ada ulang tahun sama; analog dgn tabrakan hash.")
ns = np.arange(1, 81)
plt.plot(ns, [p_sama(n) for n in ns]); plt.axhline(.5, ls="--", c="r"); plt.axvline(23, ls="--", c="g")
plt.xlabel("Jumlah orang"); plt.ylabel("P(ada yang sama)"); plt.title("Birthday Paradox")
plt.savefig("g02_birthday.png", dpi=120); plt.show()

# ======================================================================
judul("SOAL 10 - Keandalan sistem seri vs paralel")
r = .9
seri3, paralel3 = r**3, 1 - (1 - r)**3
print(f"Seri 3 komponen    : {seri3:.4f}")
print(f"Paralel 3 komponen : {paralel3:.4f}")
k = rng.random((N, 3)) < r
print(f"Simulasi seri      : {k.all(1).mean():.4f} | paralel: {k.any(1).mean():.4f}")
print("Interpretasi: redundansi (paralel) menaikkan keandalan jadi 99.9%; seri justru turun jadi 72.9%.")

# ======================================================================
judul("SOAL 11 - Hukum Bilangan Besar (lempar koin)")
lempar = rng.integers(0, 2, 10_000)
rel = np.cumsum(lempar) / np.arange(1, 10_001)
for n in (10, 100, 1000, 10000): print(f"n={n:>6}: proporsi kepala = {rel[n-1]:.4f}")
plt.semilogx(rel); plt.axhline(.5, c="r", ls="--"); plt.xlabel("n lemparan"); plt.ylabel("proporsi kepala")
plt.title("Hukum Bilangan Besar"); plt.savefig("g02_lln.png", dpi=120); plt.show()
print("Interpretasi: proporsi mendekati 0.5 saat n membesar.")

# ======================================================================
judul("SOAL 12 - Monte Carlo estimasi pi")
xy = rng.random((1_000_000, 2))
print(f"Estimasi pi = {4 * ((xy**2).sum(1) <= 1).mean():.5f} (asli {math.pi:.5f})")

# ======================================================================
judul("SOAL 13 - Peluang password acak (permutasi)")
total = 36**8
print(f"Ruang sandi 8 karakter [a-z0-9] = {total:,}")
print(f"P(tebakan benar) = {1/total:.3e}")
print(f"P(tidak ada digit sama sekali) = {(26/36)**8:.4f}")
```

```python
"""
03 - DISTRIBUSI PROBABILITAS (diskrit & kontinu) - konteks Teknik Informatika
Setiap kasus: probabilitas, E[X], Var(X), simulasi, interpretasi, + grafik.
"""
import math
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
def judul(t): print("\n" + "=" * 70 + f"\n{t}\n" + "=" * 70)

def cek(dist, sampel):
    print(f"E[X]  teori={dist.mean():.4f} | simulasi={sampel.mean():.4f}")
    print(f"Var   teori={dist.var():.4f} | simulasi={sampel.var():.4f}")

# ###################################################################
#                           DISTRIBUSI DISKRIT
# ###################################################################
judul("D1 - BERNOULLI: request berhasil dengan p=0.9")
X = stats.bernoulli(.9)
print(f"P(X=1)={X.pmf(1):.2f}, P(X=0)={X.pmf(0):.2f}"); cek(X, X.rvs(100_000, random_state=1))

judul("D2 - BINOMIAL: 20 paket data, tiap paket hilang p=0.1")
X = stats.binom(20, .1)
print(f"P(X=2)  = {X.pmf(2):.4f}")
print(f"P(X<=2) = {X.cdf(2):.4f}")
print(f"P(X>=3) = {X.sf(2):.4f}")
cek(X, X.rvs(200_000, random_state=2))
print("Interpretasi: rata-rata 2 paket hilang dari 20; peluang >=3 hilang sekitar 32%.")
bin_x = X

judul("D3 - GEOMETRIK: percobaan login sampai pertama sukses, p=0.2")
X = stats.geom(.2)
print(f"P(sukses tepat di percobaan ke-3) = {X.pmf(3):.4f}  (0.8^2*0.2)")
print(f"P(butuh > 5 percobaan)            = {X.sf(5):.4f}")
cek(X, X.rvs(200_000, random_state=3))
print("Interpretasi: rata-rata butuh 1/p = 5 percobaan.")
geo_x = X

judul("D4 - POISSON: rata-rata 4 request/detik ke server")
X = stats.poisson(4)
print(f"P(X=6)  = {X.pmf(6):.4f}")
print(f"P(X>8)  = {X.sf(8):.4f}  -> peluang server kewalahan jika kapasitas 8/detik")
print(f"Kapasitas minimum agar P(overload)<1%: {int(X.ppf(.99))} request/detik")
cek(X, X.rvs(200_000, random_state=4))
poi_x = X

judul("D5 - HIPERGEOMETRIK: 100 test case, 10 gagal, diambil 8 (tanpa pengembalian)")
X = stats.hypergeom(M=100, n=10, N=8)
print(f"P(tepat 1 gagal)  = {X.pmf(1):.4f}")
print(f"P(minimal 1 gagal)= {1 - X.pmf(0):.4f}")
cek(X, X.rvs(200_000, random_state=5))
print("Interpretasi: sampling 8 test case punya ~58% peluang mendeteksi minimal 1 kegagalan.")

judul("D6 - BINOMIAL NEGATIF: jumlah gagal sebelum 3 sukses, p=0.3")
X = stats.nbinom(3, .3)
print(f"P(gagal=5) = {X.pmf(5):.4f}")
cek(X, X.rvs(200_000, random_state=6))

judul("D7 - UNIFORM DISKRIT: pemilihan 1 dari 6 server secara acak (load balancer)")
X = stats.randint(1, 7)
print(f"P(tiap server)={X.pmf(3):.4f}, E={X.mean()}, Var={X.var():.4f}")
cek(X, X.rvs(200_000, random_state=7))

# ---- grafik diskrit
fig, ax = plt.subplots(2, 2, figsize=(11, 8))
for a, (d, k, t) in zip(ax.ravel(), [(bin_x, np.arange(0, 12), "Binomial(20,0.1)"),
                                     (geo_x, np.arange(1, 25), "Geometrik(0.2)"),
                                     (poi_x, np.arange(0, 15), "Poisson(4)"),
                                     (stats.hypergeom(100, 10, 8), np.arange(0, 6), "Hipergeometrik(100,10,8)")]):
    a.bar(k, d.pmf(k), alpha=.7); a.set_title(t); a.set_xlabel("x"); a.set_ylabel("P(X=x)")
plt.tight_layout(); plt.savefig("g03_diskrit.png", dpi=120); plt.show()

# ###################################################################
#                           DISTRIBUSI KONTINU
# ###################################################################
judul("K1 - UNIFORM KONTINU: waktu tunggu antrian 0-10 detik")
X = stats.uniform(0, 10)
print(f"P(X>7) = {X.sf(7):.2f} | P(2<X<5) = {X.cdf(5)-X.cdf(2):.2f}")
cek(X, X.rvs(200_000, random_state=8))

judul("K2 - EKSPONENSIAL: rata-rata 5 menit antar kedatangan job")
X = stats.expon(scale=5)
print(f"P(X>8)          = {X.sf(8):.4f}")
print(f"P(X>8 | X>3)    = {X.sf(8)/X.sf(3):.4f}  vs P(X>5)={X.sf(5):.4f}  -> sifat tanpa memori")
print(f"Median          = {X.median():.4f}")
cek(X, X.rvs(200_000, random_state=9))
exp_x = X

judul("K3 - NORMAL: waktu respons API ~ N(200 ms, 30 ms)")
X = stats.norm(200, 30)
print(f"P(X>250)        = {X.sf(250):.4f}")
print(f"P(170<X<230)    = {X.cdf(230)-X.cdf(170):.4f}  (aturan 68%)")
print(f"Persentil ke-95 = {X.ppf(.95):.2f} ms  (SLA p95)")
print(f"z-score 250 ms  = {(250-200)/30:.3f}")
cek(X, X.rvs(200_000, random_state=10))
nor_x = X

judul("K4 - GAMMA: total waktu 3 job berurutan, masing-masing Exp(rata-rata 2 jam)")
X = stats.gamma(a=3, scale=2)
print(f"P(total <= 4 jam) = {X.cdf(4):.4f}")
tot = rng.exponential(2, (200_000, 3)).sum(1)
print(f"Simulasi          = {(tot <= 4).mean():.4f}")
cek(X, tot)
gam_x = X

judul("K5 - WEIBULL: umur hardisk (shape=1.5, skala=1000 jam)")
X = stats.weibull_min(c=1.5, scale=1000)
print(f"P(awet > 1500 jam) = {X.sf(1500):.4f}")
print(f"Rata-rata umur     = {X.mean():.2f} jam")
cek(X, X.rvs(200_000, random_state=11))
wei_x = X

judul("K6 - CHI-SQUARE & t: nilai kritis uji statistik")
print(f"chi2 df=5, alpha=.05 -> kritis = {stats.chi2(5).ppf(.95):.4f}")
print(f"t    df=10, alpha=.05 (dua arah) -> kritis = {stats.t(10).ppf(.975):.4f}")
print(f"z    alpha=.05 (dua arah)        -> kritis = {stats.norm.ppf(.975):.4f}")

# ---- grafik kontinu
fig, ax = plt.subplots(2, 3, figsize=(15, 8))
for a, (d, xs, t) in zip(ax.ravel(), [
        (stats.uniform(0, 10), np.linspace(-1, 11, 300), "Uniform(0,10)"),
        (exp_x, np.linspace(0, 30, 300), "Eksponensial(mean 5)"),
        (nor_x, np.linspace(100, 300, 300), "Normal(200,30)"),
        (gam_x, np.linspace(0, 25, 300), "Gamma(3, 2)"),
        (wei_x, np.linspace(0, 3000, 300), "Weibull(1.5, 1000)"),
        (stats.chi2(5), np.linspace(0, 25, 300), "Chi-square(5)")]):
    a.plot(xs, d.pdf(xs), lw=2); a.fill_between(xs, d.pdf(xs), alpha=.2); a.set_title(t)
plt.tight_layout(); plt.savefig("g03_kontinu.png", dpi=120); plt.show()

# ###################################################################
#                         TOPIK LANJUTAN
# ###################################################################
judul("L1 - Pendekatan Poisson untuk Binomial (n=1000, p=0.003)")
b, p = stats.binom(1000, .003), stats.poisson(3)
for k in range(0, 6): print(f"k={k}: binom={b.pmf(k):.5f} | poisson={p.pmf(k):.5f}")
print("Interpretasi: n besar & p kecil -> Poisson(np) sangat akurat.")

judul("L2 - Pendekatan Normal untuk Binomial (n=100, p=0.5) dengan koreksi kontinuitas")
b = stats.binom(100, .5); nrm = stats.norm(50, 5)
print(f"Eksak P(X<=55)          = {b.cdf(55):.4f}")
print(f"Normal (tanpa koreksi)  = {nrm.cdf(55):.4f}")
print(f"Normal (koreksi +0.5)   = {nrm.cdf(55.5):.4f}")

judul("L3 - Central Limit Theorem: rata-rata n sampel Uniform(0,1)")
fig, ax = plt.subplots(1, 3, figsize=(14, 4))
for a, n in zip(ax, (1, 5, 30)):
    m = rng.random((50_000, n)).mean(1)
    a.hist(m, bins=50, density=True, alpha=.6)
    xs = np.linspace(m.min(), m.max(), 200)
    a.plot(xs, stats.norm.pdf(xs, .5, math.sqrt(1/12/n)), "r"); a.set_title(f"n = {n}")
    print(f"n={n:>2}: mean={m.mean():.4f}, std={m.std():.4f}, std teori={math.sqrt(1/12/n):.4f}")
plt.tight_layout(); plt.savefig("g03_clt.png", dpi=120); plt.show()
print("Interpretasi: walau asalnya uniform, rata-rata sampel makin mirip normal saat n bertambah.")

judul("L4 - Hubungan Poisson-Eksponensial: simulasi proses kedatangan (lambda=2/menit)")
antar = rng.exponential(1/2, 300_000)
waktu = np.cumsum(antar)
per_menit = np.bincount(waktu.astype(int))[:-1]
print(f"Rata-rata kedatangan/menit = {per_menit.mean():.4f} (teori 2)")
print(f"Var kedatangan/menit       = {per_menit.var():.4f} (teori 2, sifat Poisson mean=var)")

judul("L5 - Fitting distribusi ke data (MLE) + uji Kolmogorov-Smirnov")
data = rng.normal(200, 30, 2000)
mu, sd = stats.norm.fit(data)
ks = stats.kstest(data, "norm", args=(mu, sd))
print(f"Estimasi mu={mu:.2f}, sigma={sd:.2f} (asli 200, 30)")
print(f"KS stat={ks.statistic:.4f}, p-value={ks.pvalue:.4f}")
print("Interpretasi: p-value > 0.05 -> tidak ada bukti data menyimpang dari normal.")
plt.hist(data, bins=40, density=True, alpha=.6, label="Data")
xs = np.linspace(data.min(), data.max(), 200)
plt.plot(xs, stats.norm.pdf(xs, mu, sd), "r", label="Normal hasil fit"); plt.legend()
plt.savefig("g03_fit.png", dpi=120); plt.show()

judul("L6 - Distribusi gabungan diskrit & covariance (2 dadu: X=dadu1, Y=jumlah)")
d1 = rng.integers(1, 7, 200_000); d2 = rng.integers(1, 7, 200_000); Y = d1 + d2
print(f"Cov(X,Y) simulasi = {np.cov(d1, Y)[0,1]:.4f} (teori Var(X)=2.9167)")
print(f"Korelasi          = {np.corrcoef(d1, Y)[0,1]:.4f} (teori 0.7071)")
```
